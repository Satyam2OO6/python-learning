```python
import pandas as pd
from bokeh.plotting import figure
from bokeh.io import output_file, show

df = pd.read_csv("C:\\Users\\sdas2\\OneDrive\\Desktop\\tutorialspoint\\mapping\\volcanos.csv")

p = figure(title="Volcano Elevations")

p.scatter(
    x=df["Longitude"],
    y=df["Elevation"],
    size=8,
    marker="circle"
)

output_file("volcanoes.html")
show(p)
```
