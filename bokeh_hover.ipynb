```python
import pandas as pd

from bokeh.plotting import figure, show, output_file
from bokeh.models import ColumnDataSource, HoverTool

# Read data
df = pd.read_csv("time.csv")

# Convert to datetime
df["Start"] = pd.to_datetime(df["Start"])
df["End"] = pd.to_datetime(df["End"])

# Create data source
source = ColumnDataSource(df)

# Create figure
p = figure(
    x_axis_type="datetime",
    height=300,
    width=800,
    title="Motion Detection Intervals"
)

# Draw intervals
p.quad(
    left="Start",
    right="End",
    bottom=0,
    top=1,
    source=source
)

# Add hover tool
hover = HoverTool(
    tooltips=[
        ("Start", "@Start{%F %T}"),
        ("End", "@End{%F %T}")
    ],
    formatters={
        "@Start": "datetime",
        "@End": "datetime"
    },
    mode="mouse"
)

p.add_tools(hover)

p.yaxis.visible = False

output_file("Motion_Graph.html")
show(p)
```
