```python
import plotly.express as px

df = px.data.gapminder().query("year==2007")

fig = px.choropleth(
    df,
    locations="iso_alpha",
    color="lifeExp",
    hover_name="country",
    color_continuous_scale="Viridis",
    title="Life Expectancy by Country"
)

fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=[go.Sankey(
    node=dict(
        label=["Input", "Processing", "Output A", "Output B"]
    ),
    link=dict(
        source=[0, 1, 1],
        target=[1, 2, 3],
        value=[10, 6, 4]
    )
)])

fig.update_layout(title="Simple Sankey Diagram")

fig.show()
```

```python
import numpy as np
import plotly.express as px
import pandas as pd

frames = []

for t in range(1, 100):
    theta = np.linspace(0, 8*np.pi, 200)
    r = np.linspace(0, t/10, 200)

    x = r * np.cos(theta)
    y = r * np.sin(theta)

    temp = pd.DataFrame({
        "x": x,
        "y": y,
        "frame": t
    })

    frames.append(temp)

df = pd.concat(frames)

fig = px.line(
    df,
    x="x",
    y="y",
    animation_frame="frame",
    title="Animated Spiral"
)

fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go

n = 5000

theta = np.random.uniform(0, 8*np.pi, n)
r = theta + np.random.normal(0, 1, n)

x = r*np.cos(theta)
y = r*np.sin(theta)
z = np.random.normal(0, 2, n)

fig = go.Figure(data=[
    go.Scatter3d(
        x=x,
        y=y,
        z=z,
        mode='markers',
        marker=dict(
            size=2,
            color=r,
            colorscale='Plasma'
        )
    )
])

fig.update_layout(title="3D Galaxy Simulation")

fig.show()
```

```python
import numpy as np
import pandas as pd
import plotly.express as px

frames = []

x = np.linspace(0, 10, 300)

for t in range(50):
    y = np.sin(x + t/5)

    temp = pd.DataFrame({
        "x": x,
        "y": y,
        "frame": t
    })

    frames.append(temp)

df = pd.concat(frames)

fig = px.line(
    df,
    x="x",
    y="y",
    animation_frame="frame",
    title="Animated Sine Wave"
)

fig.show()
```
