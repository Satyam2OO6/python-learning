```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame({
    "Year": [2020, 2020, 2021, 2021, 2022, 2022],
    "Country": ["India", "USA", "India", "USA", "India", "USA"],
    "GDP": [2.8, 21.0, 3.1, 22.5, 3.5, 23.0],
    "Population": [1380, 331, 1395, 333, 1410, 335]
})

fig = px.scatter(
    df,
    x="GDP",
    y="Population",
    size="Population",
    color="Country",
    animation_frame="Year",
    size_max=60,
    title="GDP vs Population"
)

fig.show()
```
